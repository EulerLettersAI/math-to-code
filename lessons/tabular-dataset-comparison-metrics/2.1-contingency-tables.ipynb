## 1. Two categorical variables

A **contingency table** counts how often every combination of two categorical variables occurs. Consider ten observations with a three-level variable, `region`, and a two-level variable, `outcome`.

```python
#| label: example-categorical-data

import pandas as pd
from IPython.display import display

example = pd.DataFrame(
    {
        "region": [
            "North", "North", "North",
            "South", "South", "South",
            "West", "West", "West", "West",
        ],
        "outcome": [
            "Yes", "Yes", "No",
            "Yes", "No", "No",
            "Yes", "No", "No", "No",
        ],
    }
)
display(example)
```

## 2. Build and read the contingency table

Place the categories of `region` in the rows and the categories of `outcome` in the columns. Each interior cell is an observed count $O_{ab}$. The margins labeled `Total` make the row totals, column totals, and total sample size easy to read.

```python
#| label: contingency-table-example

observed = pd.crosstab(example["region"], example["outcome"])
table_with_totals = pd.crosstab(
    example["region"],
    example["outcome"],
    margins=True,
    margins_name="Total",
)
display(table_with_totals)
```

The table has the following counts:

| `region` | No | Yes | Total |
|---|---:|---:|---:|
| North | 1 | 2 | 3 |
| South | 2 | 1 | 3 |
| West | 3 | 1 | 4 |
| **Total** | **6** | **4** | **10** |

The contingency table used in the association calculation is the $3\times2$ interior table: $r=3$ `region` categories and $s=2$ `outcome` categories. The number of observations is the sum of its six interior counts, $n=10$. The displayed totals help interpretation but are not supplied as additional cells to a chi-squared calculation.

```python
#| label: contingency-table-checks

r, s = observed.shape
n = int(observed.to_numpy().sum())

assert (r, s, n) == (3, 2, 10)
print(f"r = {r}, s = {s}, n = {n}")
```

## Takeaway

A contingency table converts two categorical columns into a grid of joint counts. Its $r$ rows and $s$ columns correspond to category levels, its entries $O_{ab}$ count observed combinations, and the sum of all entries is the sample size $n$.
